# IDRAK — Live Prediction Notebook (Demo)

**Project:** IDRAK | Banking Intent Classification

This clean notebook loads the saved model pipeline and predicts new customer queries in real time.

## Load Model & Categories from Disk

```python
import joblib
from IPython.display import display, Markdown

model = joblib.load("banking_pipeline.joblib")
categories = joblib.load("categories.joblib")
```

## Live Prediction on New Customer Query
*Input any customer message below and run this cell to see human-readable output.*

```python
customer_message = "my pin is blocked"

probs = model.predict_proba([customer_message])[0]
top_idx = probs.argmax()
predicted_intent = categories[top_idx].replace("_", " ").title()
confidence = probs[top_idx] * 100

display(Markdown(f"""
### 🏦 **Customer Inquiry:**
> *"{customer_message}"*

---
### 🎯 **Model Prediction:**
We predict this customer intent is **`{predicted_intent}`** with **{confidence:.1f}%** confidence.
"""))
```
